## 3. Quick look at the data (2-3 real numbers)

Let's load the starter dataset and check some basic numbers to ensure we have enough data points and sufficient variance to analyze signals. Specifically, we'll check how many rows are available and how many pages show a clear 'down' trend.

```python
import pandas as pd

# Load the anonymized starter dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

total_rows = len(df)
declining_rows = len(df[df['trend_direction'] == 'down'])
median_impressions = df['impressions_90d'].median()

print(f"Total rows in starter dataset: {total_rows:,}")
print(f"Pages showing a 'down' trend: {declining_rows:,}")
print(f"Median 90-day impressions: {median_impressions:,.0f}")
```

## 4. Careful words: what I can and can't claim

**What I can claim:** I can report *observed* associations and *directional* trends (e.g. "pages with higher word counts tend to have higher engagement rates"). This work serves as *decision-support* to prioritize reviews.

**What I cannot claim:** I cannot claim *causal proof* (e.g. "adding 500 words will increase traffic") and I am not "predicting the Google algorithm". All findings are strictly observational based on historical search and analytics data.
